###  Checkpoint 1: Qruplar üzrə Təsviri Statistika

Dataset-i yüklədikdən sonra Turbo.az avtomobil elanları üzərində əsas biznes indikatoru olan Qiymət (`Price`) dəyişənini nəqliyyat vasitələrinin Ban növləri (`Body_type` — Sedan, SUV, Hetçbek və s.) üzrə təsviri statistik analizə cəlb etdim.

1. **Mərkəzi Meyllilik Göstəriciləri (Orta və Median):**
   - SUV/Krossover qrupunda orta qiymət Sedan modellərin orta qiymətindən nəzərəçarpan dərəcədə yüksəkdir.
   - Bütün ana ban növləri üzrə Orta dəyər (Mean) göstəricisi Median dəyərdən (Median)böyükdür. Bu, bazarda olan yüksək qiymətli lüks avtomobillərin ortalama qiyməti yuxarı çəkdiyini göstərir.

2. **Yayılma və Paylanma Forması (Standart Sapma və Skewness):**
   - Standart Sapma (Std): Qruplar daxilində dispersiya yüksəkdir, bu da eyni ban növünə aid avtomobillərin ilinə və yürüşünə görə qiymət diapazonunun geniş olduğunu göstərir.
   - Skewness (Meyllilik): Əsas ban növləri üzrə qiymət dəyişəninin skewness göstəricisi müsbətdir (Positive Right-Skewed). Paylanma sağa meylli olduğu üçün parametrik testlərdən əvvəl fərziyyələrin yoxlanması vacibdir.

```python
import pandas as pd
import numpy as np

df = pd.read_csv("turbo.az_cars.csv")

df.head()
```

```python
df['price_clean'] = df['price_x'].astype(str).str.replace(' ', '')
df['price_clean'] = pd.to_numeric(df['price_clean'], errors='coerce')

df['Price_AZN'] = np.where(
    df['currency_x'] == '$',
    df['price_clean'] * 1.7,
    df['price_clean']
)

df[['price_x', 'currency_x', 'Price_AZN']].head()
```

```python
df = df.dropna(subset=['Ban növü'])

top_5_bodies = df['Ban növü'].value_counts().head(5).index

df_top = df[df['Ban növü'].isin(top_5_bodies)].copy()

df_top['Ban növü'].value_counts()
```

```python
checkpoint1 = df_top.groupby('Ban növü')['Price_AZN'].agg(
    Say='count',
    Orta_Qiymət='mean',
    Median='median',
    Std_Sapma='std',
    Skewness=lambda x: x.skew()
).reset_index()

checkpoint1
```

###  Checkpoint 2: Biznes Sualları və Hipotezlərin Formalaşdırılması

---

####  Biznes Sualı 1: Sürətlər qutusunun tipi (Avtomat vs Mexaniki) avtomobilin qiymətinə təsir edirmi?
İkinci əl avtomobil bazarında avtomat transmissiyalı avtomobillərin mexaniki transmissiyaya nəzərən daha baha satıldığı ehtimal olunur. Bu fərqin təsadüfi olub-olmadığını yoxlamaq üçün t-test aparılacaqdır.

* **Null Hipotezi ($H_0$):** Avtomat və Mexaniki sürətlər qutusuna malik avtomobillərin orta satış qiymətləri arasında statistiki baxımdan əhəmiyyətli fərq yoxdur ($\mu_{\text{Avtomat}} = \mu_{\text{Mexaniki}}$).
* **Alternativ Hipotez ($H_1$):** Avtomat və Mexaniki sürətlər qutusuna malik avtomobillərin orta satış qiymətləri arasında statistiki baxımdan əhəmiyyətli fərq vardır ($\mu_{\text{Avtomat}} \neq \mu_{\text{Mexaniki}}$).

---

#### 🛠️Biznes Sualı 2: SUV (Offroader) ban növünün orta qiyməti Sedan modellərindən statistik olaraq daha yüksəkdirmi?
Bazar tendensiyasına əsasən SUV modellərinin daha baha olduğu görünür. Bu ehtimalı iki müstəqil seçmə üzrə yoxlayırıq.

* **Null Hipotezi ($H_0$):** SUV və Sedan ban növünə aid avtomobillərin orta qiymətləri arasında əhəmiyyətli fərq yoxdur ($\mu_{\text{SUV}} = \mu_{\text{Sedan}}$).
* **Alternativ Hipotez ($H_1$):** SUV ban növünə aid avtomobillərin orta qiyməti Sedan ban növünə aid avtomobillərin orta qiymətindən əhəmiyyətli dərəcədə yuxarıdır ($\mu_{\text{SUV}} > \mu_{\text{Sedan}}$).

```python
import scipy.stats as stats

auto_prices = df[df['Sürətlər qutusu'] == 'Avtomat']['Price_AZN'].dropna()
manual_prices = df[df['Sürətlər qutusu'] == 'Mexaniki']['Price_AZN'].dropna()

t_stat1, p_val1 = stats.ttest_ind(auto_prices, manual_prices, equal_var=False)

print("=== BİZNES SUALI 1: Sürətlər Qutusu (Avtomat vs Mexaniki) ===")
print(f"Avtomat Orta Qiymət: {auto_prices.mean():.2f} AZN")
print(f"Mexaniki Orta Qiymət: {manual_prices.mean():.2f} AZN")
print(f"P-value: {p_val1:.4e}")

if p_val1 < 0.05:
    print("NƏTİCƏ: H0 rədd edilir (p < 0.05). Avtomat və Mexaniki maşınların qiyməti arasında statistik olaraq əhəmiyyətli fərq VAR.")
else:
    print("NƏTİCƏ: H0 rədd edilə bilmir (p >= 0.05). Fərq statistik olaraq əhəmiyyətli deyildir.")
```

```python
suv_prices = df[df['Ban növü'] == 'Offroader / SUV, 5 qapı']['Price_AZN'].dropna()
sedan_prices = df[df['Ban növü'] == 'Sedan']['Price_AZN'].dropna()

t_stat2, p_val2 = stats.ttest_ind(suv_prices, sedan_prices, equal_var=False, alternative='greater')

print("=== BİZNES SUALI 2: Ban Növü (SUV vs Sedan) ===")
print(f"SUV Orta Qiymət: {suv_prices.mean():.2f} AZN")
print(f"Sedan Orta Qiymət: {sedan_prices.mean():.2f} AZN")
print(f"P-value: {p_val2:.4e}")

if p_val2 < 0.05:
    print("NƏTİCƏ: H0 rədd edilir (p < 0.05). SUV-lərin orta qiyməti Sedan-lardan statistik olaraq əhəmiyyətli dərəcədə YÜKSƏKDİR.")
else:
    print("NƏTİCƏ: H0 rədd edilə bilmir (p >= 0.05). SUV və Sedan qiymətləri arasında əhəmiyyətli üstünlük təsdiqlənmədi.")
```

## Checkpoint 3: Data Tipinə Uyğun Statistik Testlərin Seçilməsi və Analizi
Bu mərhələdə Turbo.az dataset-ində olan müxtəlif dəyişən tipləri nəzərə alınaraq 3 əsas statistik test seçdim və tətbiq etdim.

1. Two-Sample Welch's T-Test
Məqsəd: Avtomobilin sürətlər qutusu tipinin (Avtomat vs Mexaniki) satış qiymətinə təsirini ölçmək.

Dəyişən Tipləri: Kəsilməz (Qiymət) vs 2 Qruplu Kategorial (Sürətlər qutusu).

Niyə Welch's T-Test?: Qruplar üzrə qiymət dispersiyaları bərabər olmadığı üçün standart T-Test əvəzinə Welch's T-Test istifadə olundu.

2. One-Way ANOVA Testi
Məqsəd: Ən çox təkrarlanan Top 5 Ban növünün (Sedan, SUV, Hetçbek, Liftbek, Universal) orta qiymətləri arasında statistik olaraq əhəmiyyətli fərqin olub-olmadığını yoxlamaq.

Dəyişən Tipləri: Kəsilməz (Qiymət) vs 3-dən Çox Qruplu Kategorial (Ban növü).

Niyə ANOVA?: İkidən çox müstəqil qrupun orta göstəricilərini eyni anda müqayisə etmək üçün One-Way ANOVA testi seçildi.

3. Chi-Square Müstəqillik Testi
Məqsəd: Avtomobilin ban növü ilə sürətlər qutusu tipi arasında statistik asılılığın olub-olmadığını müəyyən etmək.

Dəyişən Tipləri: Kategorial (Ban növü) vs Kategorial (Sürətlər qutusu).

Niyə Chi-Square?: Hər iki dəyişən kategorial göstərici olduğu üçün onlar arasındakı asılılığı ölçmək üçün Chi-Square testi seçildi.

```python
import scipy.stats as stats
import pandas as pd

# 1. Two-Sample T-Test (Welch's T-Test)
auto_prices = df[df['Sürətlər qutusu'] == 'Avtomat']['Price_AZN'].dropna()
manual_prices = df[df['Sürətlər qutusu'] == 'Mexaniki']['Price_AZN'].dropna()

t_stat, p_val_t = stats.ttest_ind(auto_prices, manual_prices, equal_var=False)

print("=== 1. WELCH'S T-TEST NƏTİCƏSİ ===")
print(f"T-statistikası: {t_stat:.4f}")
print(f"p-value: {p_val_t:.4e}\n")

# 2. One-Way ANOVA Testi
body_types = df_top['Ban növü'].unique()
group_prices = [df_top[df_top['Ban növü'] == b]['Price_AZN'].dropna() for b in body_types]

f_stat, p_val_anova = stats.f_oneway(*group_prices)

print("=== 2. ONE-WAY ANOVA TESTİ NƏTİCƏSİ ===")
print(f"F-statistikası: {f_stat:.4f}")
print(f"p-value: {p_val_anova:.4e}\n")

# 3. Chi-Square Müstəqillik Testi
contingency_table = pd.crosstab(df_top['Ban növü'], df_top['Sürətlər qutusu'])
chi2_stat, p_val_chi2, dof, expected = stats.chi2_contingency(contingency_table)

print("=== 3. CHI-SQUARE TESTİ NƏTİCƏSİ ===")
print(f"Chi2-statistikası: {chi2_stat:.4f}")
print(f"p-value: {p_val_chi2:.4e}")
```

## Checkpoint 4: p-dəyərinin və Etibar İntervalının Dərin İnterpretasiyası

Bu mərhələdə yalnız p-value göstəricisinə (p < 0.05) söykənməyərək, statistik əhəmiyyətliliklə biznes əhəmiyyətliliyi arasındakı fərq tahlil olunmuşdur. Fərqin böyüklüyünü ölçmək üçün 95%-lik Etibar İntervalları (Confidence Interval - CI) hesablanmışdır.

### 1. Avtomat vs Mexaniki Müqayisəsi
* p-value göstəricisinin həddindən artıq kiçik olması (2.25e-133) böyük seçmə həcmi ilə də əlaqədardır.
* Mərkəzi Qiymət Fərqi: Avtomat maşınlar Mexaniki maşınlardan ortalama ~28,554 AZN bahadır.
* 95% Etibar İntervalı: [26,380 AZN, 30,728 AZN].
* Dərin İnterpretasiya: Biz 95% əminliklə deyə bilərik ki, bütün ikinci əl bazarında avtomat transmissiyalı maşınlar mexaniki olanlardan minimum 26,380 AZN, maksimum 30,728 AZN daha bahadır. Bu interval sıfırı (0) əhatə etmədiyi üçün fərq təkcə statistik deyil, eyni zamanda kəskin BİZNES ƏHAMİYYƏTİNƏ malikdir.

### 2. SUV vs Sedan Müqayisəsi
* Mərkəzi Qiymət Fərqi: SUV modelləri Sedan-lardan ortalama ~30,612 AZN bahadır.
* 95% Etibar İntervalı: [26,890 AZN, 34,334 AZN].
* Dərin İnterpretasiya: 95% etibar intervalı 26,890 AZN ilə 34,334 AZN aralığındadır. İntervalın sıfırdan kifayət qədər uzaq olması SUV ban növünün bazarda yaratdığı real dəyər üstünlüyünü sübut edir.

```python
import numpy as np
import scipy.stats as stats

auto_prices = df[df['Sürətlər qutusu'] == 'Avtomat']['Price_AZN'].dropna()
manual_prices = df[df['Sürətlər qutusu'] == 'Mexaniki']['Price_AZN'].dropna()

res_1 = stats.ttest_ind(auto_prices, manual_prices, equal_var=False)
ci_1 = res_1.confidence_interval(confidence_level=0.95)
mean_diff_1 = auto_prices.mean() - manual_prices.mean()

print("=== 1. AVTOMAT VS MEXANİKİ ===")
print(f"Orta Qiymət Fərqi: {mean_diff_1:.2f} AZN")
print(f"95% Etibar İntervalı: [{ci_1.low:.2f}, {ci_1.high:.2f}] AZN\n")

suv_prices = df[df['Ban növü'] == 'Offroader / SUV, 5 qapı']['Price_AZN'].dropna()
sedan_prices = df[df['Ban növü'] == 'Sedan']['Price_AZN'].dropna()

res_2 = stats.ttest_ind(suv_prices, sedan_prices, equal_var=False)
ci_2 = res_2.confidence_interval(confidence_level=0.95)
mean_diff_2 = suv_prices.mean() - sedan_prices.mean()

print("=== 2. SUV VS SEDAN ===")
print(f"Orta Qiymət Fərqi: {mean_diff_2:.2f} AZN")
print(f"95% Etibar İntervalı: [{ci_2.low:.2f}, {ci_2.high:.2f}] AZN")
```

## Checkpoint 5: Parametrik Test Fərziyyələrinin Yoxlanması

Bu mərhələdə T-Test və ANOVA kimi parametrik statistik testlərin tələb etdiyi normallıq və dispersiya bərabərliyi fərziyyələrini yoxladım.

### 1. Normallıq Fərziyyəsi
* Məqsəd: Qiymət dəyişəninin (Price_AZN) normal paylanmaya uyğun olub-olmadığını test etmək.
* Test: D'Agostino-Pearson Normallıq Testi.
* Səbəb: Seçmə həcmi böyük olduğu üçün Shapiro-Wilk əvəzinə D'Agostino-Pearson testi istifadə olundu.

### 2. Dispersiyaların Bərabərliyi Fərziyyəsi (Homoscedasticity)
* Məqsəd: Avtomat və Mexaniki qruplarının qiymət dispersiyalarının bərabər olub-olmadığını ölçmək.
* Test: Levene Testi.
* Səbəb: Məlumatlar normal paylanmaya tam uymadığı hallarda Levene testi dispersiya bərabərliyini yoxlamaq üçün ən dözümlü (robust) metod hesab olunur.

```python
import scipy.stats as stats

auto_prices = df[df['Sürətlər qutusu'] == 'Avtomat']['Price_AZN'].dropna()
manual_prices = df[df['Sürətlər qutusu'] == 'Mexaniki']['Price_AZN'].dropna()

stat_auto, p_norm_auto = stats.normaltest(auto_prices)
stat_man, p_norm_man = stats.normaltest(manual_prices)

print("=== NORMALLIQ TESTİ (D'Agostino-Pearson) ===")
print(f"Avtomat p-value: {p_norm_auto:.4e}")
print(f"Mexaniki p-value: {p_norm_man:.4e}\n")

stat_lev, p_levene = stats.levene(auto_prices, manual_prices)

print("=== DİSPERSİYA BƏRABƏRLİYİ TESTİ (Levene Test) ===")
print(f"Levene F-statistikasi: {stat_lev:.4f}")
print(f"Levene p-value: {p_levene:.4e}")
```
